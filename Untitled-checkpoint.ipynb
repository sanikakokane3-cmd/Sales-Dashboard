### step 1 : Load the dataset

```python
import pandas as pd
```

```python
## Load the data set

data = pd.read_csv("sample_data_cleaning_project - Sample_data_cleaning_project.csv")
```

```python
data
```

## step 2 : identifing messing values

```python
print("\nmissing values count:")
print(data.isnull().sum())
```

## step 3 :Handle missing value 

```python
# Fill missing Age values with the median
data['Age'] = data['Age'].fillna(data['Age'].median())

# Fill missing Salary values with the mean
data['Salary'] = data['Salary'].fillna(data['Salary'].mean())

# Remove rows with missing Join_Date values
data = data.dropna(subset=['Join_Date'])

# Display the cleaned data
print(data)
```

## Step4 -Check and remove duplicate roes

```python
data = data.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
data = data.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
print (data)
```

```python
data=data.drop_duplicates(subset=['Name','Age'])
print(data)
```

```python
print (data.duplicated().sum())
```

##Step5 : handle outliers

```python

# Calculate IQR for Salary
Q1 = data['Salary'].quantile(0.25)
Q3 = data['Salary'].quantile(0.75)

IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Filter out outliers
data = data[
    (data['Salary'] >= lower_limit) &
    (data['Salary'] <= upper_limit)
]

print(data)
```

```python
print (data.duplicated().sum())
```

##Steps6: Convert Data type

```python
data['join_Date'] = pd.to_determine(data['join_Date'])
print(data)
```

## 


Step 7 : Encode categorical Variables

```python

data = pd.get_dummies(data, columns=['Department'], drop_first=True)

binary_columns = [col for col in data.columns if col.startswith('Department_')]

data[binary_columns] = data[binary_columns].astype(int)

print(data)
```

 ## Step9:  Save the dataset

```python
data.to_csv("cleaned_data.csv",index=False)
```
